## 4.3 Distribution

The distribution of a constant or a variable is defined as the multiplication of the constant or variable by each term in a _group_. The _group_ is placed in parenthesis. The constant or variable is placed outside the parenthesis. The multiplication is performed on each term in the group.

__Definition 4.3.1__ The <span style="color:green">__distribution__</span> of a constant or a variable over a group of terms with addition is defined in __Equation 4.3.1__.

$$
x(a + b + c + \ldots + n) = ax + bx + cx + \ldots + nx \tag{Equation 4.3.1}
$$

__Problem 4.3.1__ Distribute the constant 3 over the group $a + b + c + d$.

The solution is shown in __Equation 4.3.2__.

$$
3 \left( a + b + c + d \right) = 3a + 3b + 3c + 3d \tag{Equation 4.3.2}
$$

__Problem 4.3.4__ Simplify the expression $4x(x-2) - (5x+3)$ by distribution.

The solution is shown in __Equation 4.3.5__.

$$
\begin{align*}
&4x(x-2) - (5x+3) = 4x^2 - 8x - 5x - 3 \\ \\
&4x(x-2) - (5x+3) = 4x^2 - 13x - 3
\end{align*}
\tag{Equation 4.3.5}
$$

In the solution above, we used the distribution $-1(5x+3)= -5x - 3$. We also grouped the terms $-8x$ and $-5x$ together such that $-8x-5x = -13x$.

__Problem 4.3.5__ Expand the expression $a \left(  a^{4} + 2a^{2} +3\right)$ by distribution.

The solution is shown in __Equation 4.3.6__.

$$
\begin{align*}
&a \left(  a^{4} + 2a^{2} +3\right) = a \times a^{4} + 2 \times a \times a^{2} + 3 \times a \\ \\
&a \left(  a^{4} + 2a^{2} +3\right) = a^{1+4} + 2a^{1+2} + 3a^{1} \\ \\
&a \left(  a^{4} + 2a^{2} +3\right) = a^{5} + 2a^{3} + 3a
\end{align*}
\tag{Equation 4.3.6}
$$

We confirm the last result using `sympy`. We start by defining the symbols $a$ as a mathematical symbol using the `symbols` function and assign it to the variable `a`.

```python
# Assign the symbol a to the variable a
a = symbols('a')
```

```python
# Print Problem 4.3.5 to the screen
a * (a**4 + 2 * a**2 + 3)
```

We use the `expand` method to expand the expression. This performs the distribution.

```python
# Perform the calculation in Problem 4.3.5
(a * (a**4 + 2 * a**2 + 3)).expand()
```

__Problem 4.3.8__ Expand the expression in __Equation 4.3.14__ where $x,y \ge 0$, by distribution.

$$
\sqrt{xy} \left( \sqrt{xy} + \sqrt{y} \right)
\tag{Equation 4.3.14}
$$

The result is shown in __Equation 4.3.15__.

$$
\begin{align*}
&\sqrt{xy} \left( \sqrt{xy} + \sqrt{y} \right) \\ \\
&= \sqrt{xy} \times \sqrt{xy} + \sqrt{xy} \times \sqrt{y} \\ \\
&= \sqrt{x^{2}y^{2}} + \sqrt{xy^{2}} \\ \\
&= \sqrt{x^{2}} \sqrt{y^{2}} + \sqrt{x} \sqrt{y^{2}} \\ \\
&= xy + \sqrt{x} y
\end{align*}
\tag{Equation 4.3.15}
$$

We can only write $\sqrt{x^{2}y^{2}} = \sqrt{x^{2}} \sqrt{y^{2}}$ because $x^{2}$ and $y^{2}$ are both positive numbers. We constrained $x$ and $y$ to be positive in this problem. Even if both were negative, the result would still be the same (for the first term).

The result is confirmed in the code cell below. We start by creating the mathematical variables $x$ and $y$ assigned to the variables `x` and `y` respectively. The variables are constrained to be positive. We then use the `expand` method to expand the expression.

```python
# Creating the mathematical variable x and y
x, y = symbols('x y', positive=True)
```

```python
# Verify the result
(sqrt(x * y) * (sqrt(x * y) + sqrt(y))).expand()
```

__Problem 4.3.9__ Simplify the expression in __Equation 4.3.16__ where $x,y>0$, by distribution.

$$
\sqrt{xy^{3}} \left( \sqrt{x^{5}y} + \sqrt{xy^{7}} \right)
\tag{Equation 4.3.16}
$$

The solution follows in __Equation 4.3.17__ where we make use of radicals.

$$
\begin{align*}
&\sqrt{xy^{3}} \left( \sqrt{x^{5}y} + \sqrt{xy^{7}} \right) \\ \\
= &x^{\frac{1}{2}} y^{\frac{3}{2}} \left( x^{\frac{5}{2}} y^{\frac{1}{2}} + x^{\frac{1}{2}} y^{\frac{7}{2}} \right) \\ \\
= &x^{\frac{1}{2} + \frac{5}{2}} y^{\frac{3}{2} + \frac{1}{2}} + x^{\frac{1}{2} + \frac{1}{2}} y^{\frac{3}{2} + \frac{7}{2}} \\ \\
= &x^{3} y^{2} + x y^{5}
\end{align*}
\tag{Equation 4.3.17}
$$

We solve the problem using the `expand` method in the code cell below.

```python
# Verify the result of Problem 4.3.8
(sqrt(x * y**3) * (sqrt(x**5 * y) + sqrt(x * y**7))).expand()
```

Note that if it is not stipulated that $x$ and $y$ are positive, then the combination of the bases are not possible. For example, $\sqrt{-1} \times \sqrt{-1}$ is not $\sqrt{-1 \times -1} = \sqrt{1} = \pm 1$. This is because $\sqrt{-1} = i$, the imaginary unit and $i^{2}=-1$.

Below, we recreate the mathematical variables $x$ and $y$, but allow them to be any real number.

```python
# Recreate variables x and y
x, y = symbols('x y', real=True)
```

Now we have that $\sqrt{x} \sqrt{y} \ne \sqrt{xy}$.

```python
sqrt(x) * sqrt(y)
```

Above, the bases $x$ and $y$ are not the same. Below, however, we have that the bases, $xy$, are the same. We can confirm that $\sqrt{xy} \sqrt{xy} = xy$.

```python
sqrt(x * y) * sqrt(x * y)
```

As an example we let $x=-3$ and $y=2$.

```python
sqrt(-3 * 2) * sqrt(-3 * 2)
```

The result is so because $\sqrt{-3 \times 2} = \sqrt{-6} = \sqrt{6 \left( -1 \right)} = \sqrt{6} \sqrt{-1} = \sqrt{6}i$ and $\left( \sqrt{6}i \right)^{2} = 6 i^{2} = -6$.

Note that in the case of a constant, we would evaluate the constant as given and $\sqrt{\left( -3 \right)^{2}} = \sqrt{9}$.

```python
sqrt((-3)**2)
```

__Problem 4.3.10__ Simplify the expression in __Equation 4.3.18__ by distribution.

$$
\left( 3x + 4 \right) \left( 2x - 1 \right)
\tag{Equation 4.3.18}
$$

The result is shown in __Equation 4.3.19__.

$$
\begin{align*}
&\left( 3x + 4 \right) \left( 2x - 1 \right) \\ \\
= &3x \left( 2x - 1 \right) + 4 \left( 2x - 1 \right) \\ \\
= &3x \cdot 2x - 3x \cdot 1 + 4 \cdot 2x - 4 \cdot 1 \\ \\
= &6x^{2} - 3x + 8x - 4 \\ \\
= &6x^{2} + 5x - 4
\end{align*}
\tag{Equation 4.3.19}
$$

We note that each of the terms $3x$ and $4$ in the first set of parentheses are distributed over each of the terms $2x$ and $-1$ in the second set of parentheses. The signs are important. The $3x$ is distributed over $2x$ and $-1$ and the $4$ is distributed over $2x$ and $-1$.

We verify the result using the `expand` method in `sympy`.

```python
# Verify the result of Problem 4.3.9
((3 * x + 4) * (2 * x - 1)).expand()
```

__Problem 4.3.13__ Simplify the expression in __Equation 4.3.24__ by distribution.

$$
\left( a + b \right)^{3}
\tag{Equation 4.3.24}
$$

The solution is shown in __Equation 4.3.25__ where we start by writing the term in parenthesis three times.

$$
\begin{align*}
&\left( a + b \right)^{3} \\ \\
= &\left( a + b \right) \left( a + b \right) \left( a + b \right) \\ \\
= &\left( a + b \right) \left( a^{2} + ab + ba + b^{2} \right) \\ \\
= &a \left( a^{2} + ab + ba + b^{2} \right) + b \left( a^{2} + ab + ba + b^{2} \right) \\ \\
= &a \cdot a^{2} + a \cdot ab + a \cdot ba + a \cdot b^{2} + b \cdot a^{2} + b \cdot ab + b \cdot ba + b \cdot b^{2} \\ \\
= &a^{3} + a^{2}b + aba + ab^{2} + ba^{2} + bab + b^{2}a + b^{3} \\ \\
= &a^{3} + a^{2}b + a^{2}b + ab^{2} + a^{2}b + ab^{2} + ab^{2} + b^{3} \\ \\
= &a^{3} + 3a^{2}b + 3ab^{2} + b^{3}
\end{align*}
\tag{Equation 4.3.25}
$$

In __Equation 4.3.27__ we consider the last two groups of terms first, completed that distribution, and then distributed the first group of terms. The result is confirmed in the code cell below.

```python
# Expand the expression in Problem 4.3.13 using the expand method
((a + b)**3).expand()
```
